```python
import numpy as np
import pandas as pd
```

```python
from sktime.forecasting.arima import AutoARIMA
from sktime.forecasting.base import ForecastingHorizon
```

```python
df = pd.read_csv('AirtrafficA4_cleaned.csv', index_col=0)
df.index = pd.to_datetime(df.index).to_period('M')
df
```

```python
y = df['PASSENGERS CARRIED']
```

```python
forecaster = AutoARIMA()
```

```python
fh = ForecastingHorizon(np.arange(1, 13), is_relative=True)
```

```python
y_pred = forecaster.fit_predict(y, fh=fh)
```

```python
df_pred = pd.DataFrame(y_pred)
df_pred.index = df_pred.index.to_timestamp().strftime('%Y %b').str.upper()
df_pred = df_pred.rename_axis('YEAR_MONTH')
df_pred.to_csv('predictions.csv')
```
